```python
import tensorflow as tf
from tensorflow.keras.applications import DenseNet201
from tensorflow.keras.utils import to_categorical
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix

# Data Loading and Preprocessing
def load_and_preprocess_data(dataset_name):
    if dataset_name == 'mnist':
        (x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
        x_train = x_train[..., tf.newaxis]
        x_test = x_test[..., tf.newaxis]
    elif dataset_name == 'fashion_mnist':
        (x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
        x_train = x_train[..., tf.newaxis]
        x_test = x_test[..., tf.newaxis]
    elif dataset_name == 'cifar10':
        (x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()
    elif dataset_name == 'cifar100':
        (x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar100.load_data()
    else:
        raise ValueError("Unknown dataset")

    x_train = x_train.astype('float32') / 255.0
    x_test = x_test.astype('float32') / 255.0

    # Resize images to 32x32 for DenseNet-201
    if dataset_name in ['mnist', 'fashion_mnist']:
        x_train = tf.image.resize(x_train, [32, 32])
        x_test = tf.image.resize(x_test, [32, 32])
        x_train = tf.image.grayscale_to_rgb(x_train)
        x_test = tf.image.grayscale_to_rgb(x_test)

    y_train = to_categorical(y_train)
    y_test = to_categorical(y_test)

    return x_train, y_train, x_test, y_test

# Define DenseNet-201 Model
def create_densenet201_model(input_shape, num_classes):
    base_model = DenseNet201(weights=None, include_top=False, input_shape=input_shape)
    x = base_model.output
    x = tf.keras.layers.GlobalAveragePooling2D()(x)
    x = tf.keras.layers.Dense(1024, activation='relu')(x)
    x = tf.keras.layers.Dense(num_classes, activation='softmax')(x)
    model = tf.keras.models.Model(inputs=base_model.input, outputs=x)
    return model

# Visualize Random Data Samples
def visualize_random_samples(x_data, y_data, num_classes, dataset_name):
    plt.figure(figsize=(10, 5))
    for i in range(10):
        plt.subplot(2, 5, i+1)
        plt.imshow(x_data[i])
        plt.title(f"Label: {np.argmax(y_data[i])}")
        plt.axis('off')
    plt.suptitle(f"Random Samples from {dataset_name}")
    plt.show()

# Main Function
def main():
    datasets = ['mnist', 'fashion_mnist', 'cifar10', 'cifar100']
    metrics = []

    for dataset in datasets:
        print(f"\nStarting training for {dataset}...")

        # Load and preprocess data
        x_train, y_train, x_test, y_test = load_and_preprocess_data(dataset)
        num_classes = y_train.shape[1]
        input_shape = x_train.shape[1:]

        # Visualize random samples
        visualize_random_samples(x_train, y_train, num_classes, dataset)

        # Create and compile model
        model = create_densenet201_model(input_shape, num_classes)
        model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

        # Train model
        history = model.fit(x_train, y_train, epochs=10, batch_size=32, validation_split=0.2)

        # Evaluate and store results
        test_loss, test_acc = model.evaluate(x_test, y_test, verbose=2)
        print(f"Test accuracy for DenseNet-201 on {dataset}: {test_acc}")

        # Predict and compute metrics
        y_pred = model.predict(x_test)
        y_pred_classes = y_pred.argmax(axis=-1)
        y_true_classes = y_test.argmax(axis=-1)

        report = classification_report(y_true_classes, y_pred_classes, output_dict=True)
        precision = report['accuracy']
        recall = report['accuracy']
        f1_score = report['accuracy']
        loss = test_loss

        metrics.append({
            'Dataset': dataset,
            'Accuracy': test_acc,
            'Precision': precision,
            'Recall': recall,
            'F1 Score': f1_score,
            'Loss': loss
        })

        # Confusion Matrix
        cm = confusion_matrix(y_true_classes, y_pred_classes)
        plt.figure(figsize=(10, 7))
        sns.heatmap(cm, annot=True, fmt='d')
        plt.title(f'Confusion Matrix for DenseNet-201 on {dataset}')
        plt.xlabel('Predicted')
        plt.ylabel('True')
        plt.show()

        print(f"Finished training for {dataset}.\n")

    # Create and display metrics table
    metrics_df = pd.DataFrame(metrics)
    print("Metrics Table:")
    print(metrics_df)

if __name__ == "__main__":
    main()
```
